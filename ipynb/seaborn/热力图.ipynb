```python
import seaborn as sns
import matplotlib.pyplot as plt
flights=sns.load_dataset('flights')
flights.head()
```

```python
data=flights.pivot(index='month',columns='year',values='passengers')
sns.heatmap(data,annot=True,fmt='d',cmap='Reds',cbar=True,vmin=100,vmax=600,linewidths=0.5,linecolor='white')#annot=True显示数值，fmt='d'整数，cmap颜色，cbar颜色条，vmin,vmax颜色范围，linewidths线宽，linecolor线颜色
plt.show()
```
