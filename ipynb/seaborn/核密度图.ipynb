```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset("tips")
sns.kdeplot(data=tips, x="total_bill", y="tip",hue='time', fill=True,multiple="stack",alpha=0.5)#hue='time'表示按照time列进行分组，fill=True表示填充曲线，multiple="stack"表示堆叠显示，alpha=0.5表示透明度
plt.show()
```

```python
sns.displot(data=tips, x="total_bill", y="tip",col='time', kind="kde", fill=True, alpha=0.5)
plt.show()
```
