```python
from matplotlib import pyplot as plt
import seaborn as sns
tip=sns.load_dataset('tips')
sns.displot(data=tip,x='total_bill')
sns.histplot(data=tip,x='total_bill')
```

```python
sns.displot(data=tip,x='total_bill',hue='sex',kde=True,multiple='dodge')
#multiple=layer/dodge/stack/fill(层叠/并列/堆叠/百分比堆叠)
```

```python
sns.displot(data=tip,x='total_bill',hue='sex',kde=True,multiple='dodge',color='skyblue',edgecolor='black',linewidth=1)
```

```python
sns.displot(data=tip,x='total_bill',hue='sex',kde=True,multiple='dodge',color='skyblue',edgecolor='black',linewidth=1,col='time')#col方法一般displot可用
```
