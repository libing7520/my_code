```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset("tips")
sns.lmplot(x="total_bill", y="tip", data=tips,scatter_kws={"s": 30, "color": "g", "alpha": 0.5},line_kws={"color": "r", "alpha": 0.8})#lowess=True表示使用局部加权回归，robust=True表示使用稳健回归，scatter_kws和line_kws分别设置散点图和回归线的样式。
plt.title("回归图")
plt.show()
```
