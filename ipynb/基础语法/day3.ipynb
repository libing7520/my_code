```python
score=int(input('请输入学生的成绩：'))
grade=''
if 90<=score<=100:
    grade='A'
elif 80<=score<90:
    grade='B'
elif 70<=score<80:
    grade='C'
elif 60<=score<70:
    grade='D'
else:    grade='E'
print('成绩为{0}','等级为{1}'.format(score,grade))  
```

```python
empnum=0
salaysum=0
salays=[]
while True:
    salary=int(input('请输入第{0}个员工的工资：'.format(empnum+1)))
    if salary<0:
        print('输入错误，程序结束')
        break
    else:
        print('录入成功')
        salaysum+=salary
        empnum+=1
        salays.append(salary)
print('员工总数为：{0}'.format(empnum))
print('员工工资列表为：{0}'.format(salays))
print('员工工资总和为：{0}'.format(salaysum))
print('员工平均工资为：{0}'.format(salaysum/empnum))
```

```python
#使用zip()并行迭代
names=['Tom','Jack','Rose']
ages=[20,21,22]
for name,age in zip(names,ages):
    print('{0}的年龄是{1}'.format(name,age))
```
