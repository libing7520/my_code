```python
#计算两点之间的距离
import turtle
x1,y1=100,100
x2,y2=200,200
distance=((x2-x1)**2+(y2-y1)**2)**0.5
print(distance)
#海龟绘图
turtle.penup()
turtle.goto(x1,y1)
turtle.goto(x2,y2)
turtle.pendown()
```

```python
#运算符
a=5
b=2
print(a==b) #相等
print(a!=b) #不相等
print(a> b) #大于
print(a< b) #小于
```

```python
#逻辑运算符
a=True
b=False
print(a and b) #逻辑与
print(a or b) #逻辑或
print(not a) #逻辑非
```

```python
#位运算符
a=5 #二进制：0101
b=3 #二进制：0011
print(a & b) #按位与：0001
print(a | b) #按位或：0111
print(a ^ b) #按位异或：0110
print(~a) #按位取反：1010
print(a << 1) #左移：1010
print(a >> 1) #右移：0010
```

```python
#同一运算符
a=5
b=5
print(a is b) #True，因为小整数被缓存了
c=1000
d=1000
print(c is d) #False，因为大整数没有被缓存
```

```python
#成员运算符
a=[1,2,3,4,5]
print(3 in a) #True
print(6 in a) #False
print(3 not in a) #False
print(6 not in a) #True
```

```python
#字符串切片
a='Hello World'
print(a[0]) #第一个字符
print(a[-1]) #最后一个字符
print(a[0:5]) #前五个字符
print(a[6:]) #从第六个字符到末尾
print(a[:5]) #从开头到第五个字符
```

```python
#可变字符串
import io
a='abcdefgh'
b=io.StringIO(a) #创建一个可变字符串对象
b.seek(3)
b.write('xxx')
print(b.getvalue()) #获取当前内容
```

```python
#列表
a=[1,2,'Ture',3.14]
print(a)
b=list(range(10)) #创建一个0-9的整数序列
print(b)
```

```python
a=[x*2 for x in range(5)] #列表推导式
b=[x*2 for x in range(100) if x%9==0] #列表推导式带条件
print(a,b)
```

```python
#列表元素的增加
a=[20,40]
a.append(80)
print(a)
a.insert(1,30) #在索引1的位置插入30
print(a)
a.extend([60,70]) #在列表末尾添加多个元素
print(a)
```

```python
#列表元素的删除
a=[10,20,30,40,50]
del a[0] #删除索引0的元素
print(a)
a.remove(30) #删除第一个出现的30
print(a)
a.pop(1) #删除索引1的元素
print(a)
a.clear() #清空列表
print(a)
```

```python
#列表元素的修改
a=[10,20,30,40,50]
a[0]=15 #修改索引0的元素为15
print(a)
a[1:3]=[25,35] #修改索引1到2的元素为25和35
print(a)
```

```python
#列表元素的查询
a=[10,20,30,40,50]
print(a[0]) #访问索引0的元素
print(a[-1]) #访问最后一个元素
print(a[1:4]) #访问索引1到3的元素
print(a.index(30)) #返回30的索引
print(a.index(40,2,4)) #从索引2到4开始查找40的索引
print(a.count(20)) #返回20出现的次数
print(25 in a) #检查25是否在列表中
```
