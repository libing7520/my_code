```python
class Student:
  def __init__(self,name,score):
    self.name = name    #增加name属性
    self.score = score  #增加score属性

  def say_score(self):
    self.age = 18   #增加age属性
    print("{0}的分数是{1}".format(self.name,self.score))

s1 = Student("张三",80)
s1.say_score()
print(s1.age)
s1.salary = 3000    #s1对象增加salary属性

s2 = Student("李四",90)
s2.say_score()
print(s2.age)
```

```python
#实例属性和类属性
class Student:
    school = "ABC University"  #类属性
    def __init__(self, name, age):
        self.name = name  #实例属性
        self.age = age    #实例属性
s1 = Student("Alice", 20)
s2 = Student("Bob", 22)
print(s1.school)  #访问类属性
print(s2.school)  #访问类属性
print(Student.school)  #访问类属性
s1.school = "XYZ University"  #修改s1的school属性
print(s1.school)  #访问s1的school属性
print(s2.school)  #访问s2的school属性
print(Student.school)  #访问类属性
```

```python
#类方法、静态方法、实例方法
class Student:
    school = "ABC University"  #类属性
    def __init__(self, name, age):
        self.name = name  #实例属性
        self.age = age    #实例属性 
    def instance_method(self):
        print(f"这是一个实例方法，访问实例属性：{self.name}, {self.age}")
    @classmethod
    def class_method(cls):
        print(f"这是一个类方法，访问类属性：{cls.school}")
    @staticmethod
    def static_method():
        print("这是一个静态方法，不访问任何属性")
s1 = Student("Alice", 20)
s1.instance_method()  #调用实例方法
Student.class_method()  #调用类方法
Student.static_method()  #调用静态方法
```
