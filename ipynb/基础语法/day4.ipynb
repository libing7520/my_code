```python
#函数的定义和调用
def add(a, b):
    return a + b    
result = add(3, 5)
print(result)
```

```python
def print_max(a,b):
    if a > b:
        print(a, 'is maximum')
    elif a == b:
        print(a, 'is equal to', b)
    else:
        print(b, 'is maximum')
print_max(3, 5)
print_max(7, 7)
```

```python
#返回值与参数
def max(a, b):
    if a > b:
        return a
    elif a == b:
        return 'equal'
    else:
        return b
result = max(3, 5)
print(result)
result = max(7, 7)
print(result)
```

```python
#全局变量和局部变量
x = 10  # 全局变量
def foo():
    x = 5  # 局部变量
    print("Inside foo, x =", x)
foo()  # 输出: Inside foo, x = 5
print("Outside foo, x =", x)  # 输出: Outside foo, x = 10
```
