```python
#图形嵌套
from matplotlib import pyplot as plt
plt.rcParams['font.sans-serif']=['SimHei']
fig, ax1 = plt.subplots(figsize=(10,5))
days=[1,2,3,4,5,6,7]
max_temperatures=[22,24,19,25,27,23,26]
min_temperatures=[13,10,12,17,10,12,13]
ax1.plot(days, max_temperatures, label='最高温度', color='red', marker='o')
ax1.set_xlabel('日期')
ax1.set_ylabel('最高温度 (°C)', color='red')
ax1.tick_params(axis='y', labelcolor='red')
ax2 = ax1.twinx()
ax2.plot(days, min_temperatures, label='最低温度', color='blue', marker='o')
ax2.set_ylabel('最低温度 (°C)', color='blue ')
ax2.tick_params(axis='y', labelcolor='blue')
x_names=['周一','周二','周三','周四','周五','周六','周日']
ax1.set_xticks(days)
ax1.set_xticklabels(x_names)
ax1.tick_params(axis='x', labelrotation=45)
plt.title('一周内的气温变化')
fig.legend(loc='upper left', bbox_to_anchor=(0.1,0.9))
plt.grid()
plt.show()
```

```python
#绘制文本
fig, ax = plt.subplots(figsize=(10,5))
days=[1,2,3,4,5,6,7]
max_temperatures=[22,24,19,25,27,23,26]
ax.plot(days, max_temperatures, label='最高温度', color='red', marker='o')
ax.set_xlabel('日期')
ax.set_ylabel('最高温度 (°C)', color='red')
ax.tick_params(axis='y', labelcolor='red')
x_names=['周一','周二','周三','周四','周五','周六','周日']
ax.set_xticks(days)
ax.set_xticklabels(x_names)
ax.tick_params(axis='x', labelrotation=45)
plt.title('一周内的最高温度变化')
ax.legend(loc='upper left')
plt.grid()
for i, temp in enumerate(max_temperatures):
    ax.text(days[i], temp+0.5, f'{temp}°C', ha='center', va='bottom', fontsize=9, color='red')#设置文本位置和样式
plt.show()
plt.savefig('temperature_plot.png', dpi=300) #保存图像
```

```python
import matplotlib.pyplot as plt


species = ("苹果","蓝莓","橘子")
weight_counts = {
  "第1季度": [70, 39, 58],
  "第2季度": [82, 37, 66],
}
width = 0.2
x = range(len(species))
x1 =[_+width for _ in x]
width = 0.2
container = plt.bar(x, weight_counts["第1季度"],color='tab:red',label='第1季度',width=width)
plt.bar_label(container)
container2 = plt.bar(x1, weight_counts["第2季度"],color='tab:blue',label='第2季度',width=width)
plt.bar_label(container2)#数字·显示在柱子上
plt.xticks([_+width/2 for _ in x], species)
plt.legend(loc=1)
```

```python
import matplotlib.pyplot as plt

species = ("苹果","蓝莓","橘子")
weight_counts = {
  "第1季度": [70, 39, 58],
  "第2季度": [82, 37, 66],
}
x = range(len(species))
container = plt.bar(x, weight_counts["第1季度"],color='tab:red',label='第1季度')
plt.bar_label(container,label_type='center')
container2 = plt.bar(x, weight_counts["第2季度"],color='tab:blue',label='第2季度',bottom=weight_counts["第1季度"])#设置第二季度的柱子在第一季度的柱子上方
plt.bar_label(container2,label_type='center')
plt.xticks([_+width/2 for _ in x], species)
plt.legend(loc="upper right")
```
