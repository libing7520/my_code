#随机数组

```python
arr=np.random.randint(1,20,(5,5))
print(arr)
print(arr[arr>10])
print(np.where(arr>10,0,arr))
```

#统计函数应用

```python
arr=np.array([120,135,110,125,130,140])
print('总和;',np.sum(arr))
print('均值：',np.mean(arr))
print('方差：',np.var(arr))
print('标准差：',np.std(arr))
print('销售额最大月份：',np.argmax(arr)+1)
print('销售额最小月份：',np.argmin(arr)+1)
```

#数组拼接

```python
A=np.array([1,2,3])
B=np.array([4,5,6])
C=np.concatenate((A,B))
print(C)
D=np.reshape(C,(2,3))
print(D)
```

#唯一值与排序

```python
arr=np.array([2,1,2,3,1,4,3])
print('唯一元素：',np.unique(arr))
u_arr,counts=np.unique(arr,return_counts=True)
print('唯一元素出现次数：',counts)#numpy方法
d=[]
for i in range(len(u_arr)):
    d=d+[len(arr[arr==u_arr[i]])]
print(d)    
    #python方法
```

#综合运用

```python
money=np.array([20,25,22,30,28])
cost=np.array([15,18,16,22,28])
profit=money-cost
print('利润：',profit)
print('利润的平均值：',np.mean(profit))
print('利润的标准差：',np.std(profit))
print('利润最大月份：',np.argmax(profit)+1)
```
